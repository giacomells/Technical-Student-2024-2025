# Testing the full simulation

```python
xline = xt.Line.from_json("/Users/matteogiacomelli/Documents/Technical_Student/simulation/momentum-driven-crystal-sx/database/lhc_q20.json")
xline.particle_ref = xt.Particles(p0c=400e9, mass0=xt.PROTON_MASS_EV)
stein_args = SteinArgs(approach_side=1, nsigma_emit=3)
my_crystal = sps_crystal.copy()
my_crystal.bending_angle = 170e-6
ste = Steinbach(line=xline, stein_args=stein_args, crystal=my_crystal)
ste.create_xsuite_particles(plot=True);
```

```python
new_line = ste.line.cycle(name_first_element=ste.stein_args.crystal_loc)
```

```python
tw = new_line.twiss(method='4d')
tw_pandas = ste.parse_xsuite_twiss(tw)
tw_crystal = tw_pandas.loc[ste.stein_args.crystal_loc]
delta0 = tw_crystal['delta0_aligned']
ste.crystal.tilt = 0
```

```python
rmat_to_mst = tw.get_R_matrix(start=ste.stein_args.crystal_loc,
                               end=ste.stein_args.septum_loc)
line_from_mst = new_line.cycle(name_first_element=ste.stein_args.septum_loc)
tw_from_mst = line_from_mst.twiss(method='4d')  
rmat_to_cry = tw_from_mst.get_R_matrix(start=ste.stein_args.septum_loc,
                                        end=ste.stein_args.crystal_loc)
```

```python
# We are pretty much ready to simulate
# We want to observe the phase space at the MST now

extraction_time = 5
dpp_dturn = ste.stein_args.approach_side*ste.beam_args.dpp_fullwidth*23e-6/extraction_time
mst_x0, mst_dx = -7.5e-3, 5e-3
turns = int(2*np.abs(tw_crystal['dpp_stopband_rms']*ste.stein_args.nstopbands_dpp/dpp_dturn))
extracted = []

particles = ste.xsuite_particles.copy()

utils.track_with_mat(particles, rmat_to_mst)

for turn in range(turns):
    
    utils.track_with_mat(particles, rmat_to_cry)
    
    # At crystal
    ste.crystal.track(particles)
    
    # To mst
    utils.track_with_mat(particles, rmat_to_mst)
    
    # Energy update
    particles.delta += dpp_dturn
        
    out = particles.x < mst_x0 - mst_dx/2
    lost = np.abs(particles.x - mst_x0) < mst_dx/2
    particles.state[out] = -42
    particles.state[lost] = -1
```

```python
particles_init_mst = ste.xsuite_particles.copy()
utils.track_with_mat(particles_init_mst, rmat_to_mst)

isout = particles.state == -42
islost = particles.state < 0
nlost = np.sum(islost)
nout = np.sum(isout)

f, ax = plotters.subplots(1, 2, sharex=True, sharey=True)
factor = 1e3
ax[0].scatter(particles_init_mst.x*factor,
              particles_init_mst.px*factor, label='initial', c=particles_init_mst.state, marker='.')
ax[1].scatter(particles.x*factor, particles.px*factor,
              label='mid-spill', c=islost, marker='.')


for axis in ax:
    axis.axvline(factor*(mst_x0 - mst_dx/2), color='fuchsia',
                 label='MST 5mm blade')
    axis.axvline(factor*(mst_x0 + mst_dx/2), color='fuchsia')
    axis.legend()
    
    axis.set_ylabel('xp [mrad]')
    axis.set_xlabel('x [mm]')

f.suptitle(f'Out: {nlost}, Well out: {nout}, Portion: {nout/nlost}')
```
